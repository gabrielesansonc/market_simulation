# Two-Outcome Market Simulation Demo

This notebook shows how to use `simulate_two_outcome_market` from `market_simulatio.py`.

```python
%load_ext autoreload
%autoreload 2

import random
from pprint import pprint
from market_simulatio import simulate_two_outcome_market
```

```python
# x: number of offers sampled from a normal distribution
x_dist = lambda: random.gauss(20, 0)

# y: base probability for outcome A
y = 0.55

result = simulate_two_outcome_market(
    x_dist=x_dist,            # How many offers to simulate (sampled from distribution)
    y=y,                      # Base probability for outcome A
    bettor_std=0.10,          # How much bettors disagree around y (belief noise)
    ask_spread_mean=0.00,     # Average amount asks are above bettor fair value
    ask_spread_std=0.00,     # Variation in ask spread
    rng_seed=7,               # Fixed seed for reproducible results
)

print('Offers simulated:', result['n_offers'])
print('Base y for A:', result['base_probability_y_for_A'])
print('Implied probabilities:', result['market_implied_probabilities'])
print('Exchange profit:', result['exchange_profit'])
```
